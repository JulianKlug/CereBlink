```python
import pandas as pd
```

```python
dci_timings_path = '/Users/jk1/Library/CloudStorage/OneDrive-UniversitédeGenève/icu_research/dci_sah/data/sos_sah_data/dci_timings/dci_timings.xlsx'
```

```python
dci_timings_df = pd.read_excel(dci_timings_path)
```

```python
dci_timings_df = dci_timings_df[dci_timings_df['Date_admission'] >= '2008-01-01']
```

```python
dci_timings_df = dci_timings_df[dci_timings_df['DCI_YN_verified'] == 1]
```

```python
dci_timings_df['clinical symptoms'].isna().sum()
# drop rows with missing clinical symptoms
dci_timings_df = dci_timings_df.dropna(subset=['clinical symptoms'])
```

```python
# report n and % of rows with each clinical symptom
clinical_signs = ['decreased_consciousness', 'focal_neuro_signs',
       'raised_icp', 'suspect_TCD', 'delirium']
for sign in clinical_signs:
    n = dci_timings_df[sign].sum()
    pct = n / len(dci_timings_df) * 100
    print(f'{sign}: {n} ({pct:.1f}%)')
```
