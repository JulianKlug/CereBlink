```python
import os

import pandas as pd

from ischemia_timing.late_dci.data_sources import DCI_TIMINGS_FILE, DEFAULT_DATA_DIR
from ischemia_timing.late_dci.cohort import FIRST_INCLUDED_YEAR
```

```python
dci_timings_path = os.path.join(DEFAULT_DATA_DIR, DCI_TIMINGS_FILE)
```

```python
dci_timings_df = pd.read_excel(dci_timings_path)
```

```python
# Study period starts in 2011; 2009-2010 excluded
dci_timings_df = dci_timings_df[dci_timings_df['Date_admission'] >= f'{FIRST_INCLUDED_YEAR}-01-01']
```

```python
dci_timings_df = dci_timings_df[dci_timings_df['DCI_YN_verified'] == 1]
```

```python
dci_timings_df['clinical_symptoms_text'].isna().sum()
# drop rows with missing clinical symptoms
dci_timings_df = dci_timings_df.dropna(subset=['clinical_symptoms_text'])
```

```python
# report n and % of rows with each clinical symptom
clinical_signs = ['decreased_consciousness', 'focal_neuro_signs',
       'raised_icp', 'suspect_TCD', 'delirium']
for sign in clinical_signs:
    n = dci_timings_df[sign].sum()
    pct = n / len(dci_timings_df) * 100
    print(f'{sign}: {n} ({pct:.1f}%)')
```
