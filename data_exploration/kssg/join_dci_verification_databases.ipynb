```python
jh_df = load_encrypted_xlsx(jh_path)
geiss_df = load_encrypted_xlsx(geiss_path)
jk_df = load_encrypted_xlsx(jk_path)
```

```python
jh_df.columns
```

```python
columns_to_keep = ['SOS-CENTER-YEAR-NO.', 'Name', 'Date_birth', 'Date_Ictus',
       'Date_admission', 'Date_Discharge',
       'Date_DCI_ischemia_first_image', 'Time_DCI_ischemia_first_image',
       'Date_DCI_infarct_first_image', 'Time_DCI_infarct_first_image',
       'Comment', 'DCI_YN_verified', 'clinical symptoms']
```

```python
joined_df = jh_df[columns_to_keep] 
```

```python
dual_check_columns = ['SOS-CENTER-YEAR-NO.', 'Name', 'Date_birth', 'Date_DCI_ischemia_first_image_dual_checked',
       'Time_DCI_ischemia_first_image_dual_checked',
       'Date_DCI_infarct_first_image_dual_checked',
       'Time_DCI_infarct_first_image_dual_checked', 'Comment_dual_check',
       'DCI_YN_verified_dual_check', 'clinical symptoms']
```

```python
geiss_df = geiss_df[dual_check_columns]
jk_df = jk_df[dual_check_columns]
```

```python
# in Date_DCI_infarct_first_image_dual_checked and Date_DCI_ischemia_first_image_dual_checked replace - with .
geiss_df['Date_DCI_infarct_first_image_dual_checked'] = geiss_df['Date_DCI_infarct_first_image_dual_checked'].str.replace('-', '.')
geiss_df['Date_DCI_ischemia_first_image_dual_checked'] = geiss_df['Date_DCI_ischemia_first_image_dual_checked'].str.replace('-', '.')
```

```python
# merge geiss_df to jh_df
joined_df = pd.merge(joined_df, geiss_df, how='left', on=['SOS-CENTER-YEAR-NO.', 'Name', 'Date_birth'])
```

```python
joined_df['Date_DCI_ischemia_first_image'] = joined_df['Date_DCI_ischemia_first_image_dual_checked'].where(joined_df['Date_DCI_ischemia_first_image_dual_checked'].notna(), joined_df['Date_DCI_ischemia_first_image'])
joined_df['Time_DCI_ischemia_first_image'] = joined_df['Time_DCI_ischemia_first_image_dual_checked'].where(joined_df['Time_DCI_ischemia_first_image_dual_checked'].notna(), joined_df['Time_DCI_ischemia_first_image'])

joined_df['Date_DCI_infarct_first_image'] = joined_df['Date_DCI_infarct_first_image_dual_checked'].where(joined_df['Date_DCI_infarct_first_image_dual_checked'].notna(), joined_df['Date_DCI_infarct_first_image'])
joined_df['Time_DCI_infarct_first_image'] = joined_df['Time_DCI_infarct_first_image_dual_checked'].where(joined_df['Time_DCI_infarct_first_image_dual_checked'].notna(), joined_df['Time_DCI_infarct_first_image'])

joined_df['Comment'] = joined_df['Comment_dual_check'].where(joined_df['Comment_dual_check'].notna(), joined_df['Comment'])
joined_df['DCI_YN_verified'] = joined_df['DCI_YN_verified_dual_check'].where(joined_df['DCI_YN_verified_dual_check'].notna(), joined_df['DCI_YN_verified'])
joined_df['clinical symptoms_x'] = joined_df['clinical symptoms_y'].where(joined_df['clinical symptoms_y'].notna(), joined_df['clinical symptoms_x'])
```

```python
joined_df = joined_df.drop(columns=['Date_DCI_ischemia_first_image_dual_checked',
       'Time_DCI_ischemia_first_image_dual_checked',
       'Date_DCI_infarct_first_image_dual_checked',
       'Time_DCI_infarct_first_image_dual_checked', 'Comment_dual_check',
       'DCI_YN_verified_dual_check', 'clinical symptoms_y'])
```

```python
joined_df = pd.merge(joined_df, jk_df, how='left', on=['SOS-CENTER-YEAR-NO.', 'Name', 'Date_birth'])
```

```python
joined_df['Date_DCI_ischemia_first_image'] = joined_df['Date_DCI_ischemia_first_image_dual_checked'].where(joined_df['Date_DCI_ischemia_first_image_dual_checked'].notna(), joined_df['Date_DCI_ischemia_first_image'])
joined_df['Time_DCI_ischemia_first_image'] = joined_df['Time_DCI_ischemia_first_image_dual_checked'].where(joined_df['Time_DCI_ischemia_first_image_dual_checked'].notna(), joined_df['Time_DCI_ischemia_first_image'])

joined_df['Date_DCI_infarct_first_image'] = joined_df['Date_DCI_infarct_first_image_dual_checked'].where(joined_df['Date_DCI_infarct_first_image_dual_checked'].notna(), joined_df['Date_DCI_infarct_first_image'])
joined_df['Time_DCI_infarct_first_image'] = joined_df['Time_DCI_infarct_first_image_dual_checked'].where(joined_df['Time_DCI_infarct_first_image_dual_checked'].notna(), joined_df['Time_DCI_infarct_first_image'])

joined_df['Comment'] = joined_df['Comment_dual_check'].where(joined_df['Comment_dual_check'].notna(), joined_df['Comment'])
joined_df['DCI_YN_verified'] = joined_df['DCI_YN_verified_dual_check'].where(joined_df['DCI_YN_verified_dual_check'].notna(), joined_df['DCI_YN_verified'])
joined_df['clinical symptoms_x'] = joined_df['clinical symptoms'].where(joined_df['clinical symptoms'].notna(), joined_df['clinical symptoms_x'])
```

```python
joined_df = joined_df.drop(columns=['Date_DCI_ischemia_first_image_dual_checked',
       'Time_DCI_ischemia_first_image_dual_checked',
       'Date_DCI_infarct_first_image_dual_checked',
       'Time_DCI_infarct_first_image_dual_checked', 'Comment_dual_check',
       'DCI_YN_verified_dual_check', 'clinical symptoms'])
```

```python
# clinical symptoms_x -> clinical symptoms
joined_df = joined_df.rename(columns={'clinical symptoms_x': 'clinical symptoms'})
```
