# Two panel figure - Overall & with DCI

```python
non_DCI_data_path = '/Users/jk1/temp/bp_dci/bp_burden/rerun_14062026/restrict_to_non_dci/systole_event_counts_mrs_1y_association_after_aneurysm_secured.csv'
DCI_data_path = '/Users/jk1/temp/bp_dci/bp_burden/rerun_14062026/restrict_to_dci/systole_event_counts_mrs_1y_association_after_aneurysm_secured.csv'
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
non_DCI_df = pd.read_csv(non_DCI_data_path)
DCI_df = pd.read_csv(DCI_data_path)
```

```python
# two panel figure (two columns, one for overall, one for DCI only)
fig, axes = plt.subplots(1, 3, figsize=(13, 6), gridspec_kw={'width_ratios': [1, 1, 0.05]})

coefficient_name: str = 'correlation_coefficient'
annotations = False

shared_values = pd.concat([non_DCI_df[coefficient_name], DCI_df[coefficient_name]])
vmin = shared_values.min()
vmax = shared_values.max()

overall_heatmap = non_DCI_df.pivot_table(
    index='duration_threshold',
    columns='intensity_threshold',
    values=coefficient_name,
).reindex(index=sorted(non_DCI_df['duration_threshold'].unique(), reverse=True))

ax1 = axes[0]
sns.heatmap(
    overall_heatmap,
    annot=annotations,
    cmap='seismic',
    center=0,
    vmin=vmin,
    vmax=vmax,
    cbar=False,
    ax=ax1,
)

DCI_heatmap = DCI_df.pivot_table(
    index='duration_threshold',
    columns='intensity_threshold',
    values=coefficient_name,
).reindex(index=sorted(DCI_df['duration_threshold'].unique(), reverse=True))

ax2 = axes[1]
sns.heatmap(
    DCI_heatmap,
    annot=annotations,
    cmap='seismic',
    center=0,
    vmin=vmin,
    vmax=vmax,
    cbar=True,
    cbar_ax=axes[2],
    ax=ax2,
)

# x label
ax1.set_xlabel('Intensity threshold (mmHg)', fontsize=12)
ax2.set_xlabel('Intensity threshold (mmHg)', fontsize=12)

ax1.set_xlim(60, 200)
ax2.set_xlim(60, 200)
# xtick steps of 20
ax1.set_xticks(range(60, 201, 20))
ax2.set_xticks(range(60, 201, 20))
# xtick labels, oriented horizontally, with steps of 20
ax1.set_xticklabels(range(60, 201, 20), rotation=0)
ax2.set_xticklabels(range(60, 201, 20), rotation=0)

# y label
ax1.set_ylabel('Duration threshold (minutes)', fontsize=12)
ax2.set_ylabel('Duration threshold (minutes)', fontsize=12)
# ytick steps of 20, from 0 to 180
yticks1 = [i for i, v in enumerate(overall_heatmap.index) if v % 20 == 0]
yticks2 = [i for i, v in enumerate(DCI_heatmap.index) if v % 20 == 0]

ax1.set_yticks([i + 0.5 for i in yticks1])
ax1.set_yticklabels([overall_heatmap.index[i] for i in yticks1], rotation=0)

ax2.set_yticks([i + 0.5 for i in yticks2])
ax2.set_yticklabels([DCI_heatmap.index[i] for i in yticks2], rotation=0)

# increase space between ax1 and ax2
fig.subplots_adjust(wspace=0.3)
```
