# Nimodipine to Noradrenaline ratio

60mg - 0-4
30mg - 5-9
0 - > 10

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# create mock data for 2 patients (A/B) each receiving 2 treatments (nimo and nor) over 10 time points
patient_A = {
    'nimo': [60, 60, 60, 60, 60, 60, 60, 60, 60, 60],
    'nor': [0, 0, 5, 10, 20, 25, 20, 15, 10, 15],
    'outcome': 1
}

patient_B = {
    'nimo': [60, 60, 30, 30, 0, 0, 30, 15, 15, 15],
    'nor': [0, 0, 5, 7, 15, 15, 7, 5, 5, 0],
    'outcome': 5 
}
```

```python
# add function (nimo ^2)/(nor + 1) to the data

patient_A['f'] = np.array(patient_A['nimo'])**2 / (np.array(patient_A['nor']) + 1)
patient_B['f'] = np.array(patient_B['nimo'])**2 / (np.array(patient_B['nor']) + 1)
```

```python
patient_B
```

```python
# plot data from two patients in two subplots
fig, axes = plt.subplots(1, 2, figsize=(15, 5), sharey=True)

time =  np.arange(1, 11)
sns.scatterplot(x=time, y=patient_A['nimo'], ax=axes[0], label='nimo', color='blue')
sns.scatterplot(x=time, y=patient_A['nor'], ax=axes[0], label='nor', color='purple')
sns.lineplot(x=time, y=patient_A['f'], ax=axes[0], label='function', color='green')
axes[0].set_title('Patient A')

sns.scatterplot(x=time, y=patient_B['nimo'], ax=axes[1], label='nimo', color='blue')
sns.scatterplot(x=time, y=patient_B['nor'], ax=axes[1], label='nor', color='purple')
sns.lineplot(x=time, y=patient_B['f'], ax=axes[1], label='function', color='green')
axes[1].set_title('Patient B')

axes[0].set_ylim(0, 500)
```
